# Call 5 day / 3 hour forecast data

```python
website = 'http://api.openweathermap.org/data/2.5/forecast?lat='+str(lat)+'&lon='+str(long)+'&appid='+api_key
result = requests.get(website)
data = result.json()
print(json.dumps(data,indent=5))
```

```python
import pandas as pd
dt=[]
t=[]
p=[]
h=[]
d=[]
for i in data['list']:
    dt.append(i['dt_txt'])
    t.append(i['main']['temp'])
    p.append(i['main']['pressure'])
    h.append(i['main']['humidity'])
    d.append(i['weather'][0]['description'])
dfr={'dt_txt': dt, 'temp': t, 'pressure': p, 'humidity':h, 'description': d}
df=pd.DataFrame(dfr)
df
```
